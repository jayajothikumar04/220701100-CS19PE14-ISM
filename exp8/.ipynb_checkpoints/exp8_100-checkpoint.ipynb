```python
import cv2
import os

# Mapping characters to ASCII codes and vice versa
d = {}
c = {}
for i in range(255):
    d[chr(i)] = i
    c[i] = chr(i)

# Read the cover image (make sure the path is correct)
x = cv2.imread(r"D:\sem7\InformationSecurity&Management\Lab\exp8\img.jpg")
if x is None:
    print("Error: Image not found at the specified path.")
    exit()

rows = x.shape[0]
cols = x.shape[1]
print(f"Image dimensions: {rows} x {cols}")

# Input secret key and text
key = input("Enter key to edit(Security Key) : ")
text = input("Enter text to hide : ")

n = 0  # row index
m = 0  # column index
z = 0  # channel index (0-2 for BGR)
kl = 0  # key index
l = len(text)

for i in range(l):
    # XOR the pixel value with corresponding key character ASCII code
    x[n, m, z] = d[text[i]] ^ d[key[kl]]

    # Move indices to next pixel position and channel cycling
    n += 1
    if n >= rows:
        n = 0
        m += 1
        if m >= cols:
            print("Warning: Text too long to hide in image.")
            break

    m = m % cols
    kl = (kl + 1) % len(key)

# Save the encrypted image
cv2.imwrite("encrypted_img.jpg", x)
print("Data Hiding in Image completed successfully.")
os.startfile("encrypted_img.jpg")

# Option to extract hidden data
ch = int(input("\nEnter 1 to extract data from Image : "))
if ch == 1:
    key1 = input("\nRe-enter key to extract text : ")
    decrypt = ""

    if key == key1:
        n = 0
        m = 0
        kl = 0
        for i in range(l):
            val = x[n, m, z] ^ d[key[kl]]
            decrypt += c[val]

            n += 1
            if n >= rows:
                n = 0
                m += 1

            m = m % cols
            kl = (kl + 1) % len(key)
        print("Encrypted text was : ", decrypt)
    else:
        print("Key doesn't matched.")
else:
    print("Thank you. EXITING.")

```
